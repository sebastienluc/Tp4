#6 Affichage de signaux 1D avec Matplotlib et Plotly


```python
# Création d'un signal sinusoïdal
t = np.linspace(0, 2*np.pi, 100)
x = np.sin(t)
# Affichage avec Matplotlib
plt.plot(t, x, label="sin(t)")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.title("Signal sinusoïdal")
plt.legend()
plt.grid()
plt.show()
```

```python
# Création du signal
t = np.linspace(0, 2*np.pi, 100)
x = np.sin(t)
# Affichage avec Plotly Express
px.line(x=t, y=x, labels={"x":"Amplitude", "y":"Temps (s)"},
title="Signal sinusoïdal interactif").show()
```

```python
import pandas as pd
# Création d'un DataFrame depuis un dictionnaire
data = {
"Nom": ["Alice", "Bob", "Charlie"],
"Age": [23, 30, 27],
"Ville": ["Paris", "Lyon", "Marseille"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Génération du vecteur temps
t = np.linspace(0, 2*np.pi, 200)
# Génération d'un signal sinusoïdal
s = np.sin(t)
df_signal = pd.DataFrame({
"Temps (s)": t,
"Valeur": s,
})
# Affichage interactif
px.line(df_signal, x="Temps (s)", y="Valeur", title="Signaux sinusoïdal").show()
```

```python
# Génération de deux vecteurs temps différents
t1 = np.linspace(0, 2*np.pi, 200)
t2 = np.linspace(0, 4*np.pi, 400)
# Génération du DataFrame
df_signaux = pd.DataFrame({
"Temps (s)": np.concatenate([t1, t2]),
"Valeur": np.concatenate([np.sin(t1), np.cos(t2)]),
"Signal": ["sin(t)"]*len(t1) + ["cos(t)"]*len(t2)
})
# Affichage simultané
px.line(df_signaux, x="Temps (s)", y="Valeur", color="Signal", title="Signaux sinusoïdal et cosinusoidal").show()
```

```python
def somme(n):
    s=0
    for i in range(1,n+1):
        s=s+(1/i**2)
    return s  
somme(15)
```

```python
def indice_convergence(precision):
    cible = (np.pi ** 2) / 6
    n = 1
    somme = 0   
    while abs(cible - somme) >= precision:
        somme += 1 / (n ** 2)
        n += 1     
    return n - 1

indice_convergence(1e-4)
```

```python
def indice_convergence(precision):
    cible = np.log(2)
    n = 1
    somme = 0   
    while abs(cible - somme) >= precision:
        somme += ((-1)**(n+1))/n
        n += 1     
    return n - 1

indice_convergence(1e-4)

plt.plot(somme, n, label="approxiamation")
plt.xlabel("indice")
plt.ylabel("precision")
plt.show()

```
